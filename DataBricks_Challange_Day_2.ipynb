```python
events = spark.read.csv(
    "/Volumes/workspace/ecommerce/ecommerce_data",
    header=True,
    inferSchema=True
)

events.show(5)
events.printSchema()
```

```python
events.select(
    "event_type",
    "brand",
    "category_code",
    "price"
).show(10)
```

```python
events.filter(events.price >1000).count()
```

```python
events.groupBy("event_type").count().show()
```

```python
top_brands = (
    events
    .groupBy("brand")
    .count()
    .orderBy("count", ascending=False)
    .limit(10)
)

top_brands.show()

```

```python
events.groupBy("brand") \
      .sum("price") \
      .withColumnRenamed("sum(price)", "total_revenue") \
      .orderBy("total_revenue", ascending=False) \
      .show(10)

```

```python
from pyspark.sql.functions import hour

events.withColumn("hour", hour("event_time")) \
      .groupBy("hour") \
      .count() \
      .orderBy("hour") \
      .show()
```
